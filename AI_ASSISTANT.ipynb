```python
from google import genai

client = genai.Client(
    api_key=api_key
)
```

```python
system_prompt = """
You are an English grammar expert. Correct grammar, punctuation, and sentence structure while keeping the original meaning.
"""

while True:

    question = input("\nYou :")

    if question.lower() == "exit":
        break
    
    prompt = f"""
{system_prompt}

Student:
{question}
"""

    response = client.models.generate_content(
        model="models/gemini-3.5-flash-lite",
        contents=prompt
    )

    print("\nAssistant:")
    print(response.text)
```
